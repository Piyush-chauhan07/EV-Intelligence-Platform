```python
ev = pd.read_csv("../data/indian_ev_specs.csv", skiprows=1)

print("✅ Dataset loaded correctly!")
print("Shape:", ev.shape)
print("Columns:", ev.columns.tolist())

display(ev.head())
```

```python
print("\nMissing values:")
print(ev.isnull().sum())

print("\nDuplicate rows:")
print(ev.duplicated().sum())

print("\nData types:")
print(ev.dtypes)
```

```python
ev.describe()
```

```python
ev[["Brand", "Model", "Price"]].sort_values("Price", ascending=False)
```

```python
ev[["Brand", "Model", "Price"]].sort_values("Price").head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(ev["Range"], ev["Price"])

plt.xlabel("Range (km)")
plt.ylabel("Price (₹ Lakh)")
plt.title("EV Price vs Driving Range")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(ev["Battery"], ev["Price"])

plt.xlabel("Battery Capacity (kWh)")
plt.ylabel("Price (₹ Lakh)")
plt.title("EV Price vs Battery Capacity")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(ev["Power"], ev["Price"])

plt.xlabel("Power")
plt.ylabel("Price (₹ Lakh)")
plt.title("EV Price vs Power")

plt.show()
```

```python
ev[["Price", "Range", "Power", "Battery"]].corr()
```

```python
correlations = (
    ev[["Price", "Range", "Power", "Battery"]]
    .corr()["Price"]
    .sort_values(ascending=False)
)

print(correlations)
```
